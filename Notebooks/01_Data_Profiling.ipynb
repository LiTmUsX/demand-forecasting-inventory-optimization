```python
MIN_ORDERS_FOR_OWN_AVERAGE = 30
network_avg = orders['Unit quantity'].sum() / len(orders)

per_plant = orders.groupby('Plant Code').agg(
    order_count=('Order ID', 'count'),
    total_units=('Unit quantity', 'sum')
).reset_index()

per_plant['own_avg_order_size'] = (
    per_plant['total_units'] / per_plant['order_count']
)

print(f"Plants present in order history: {len(per_plant)}")
print(f"Missing from lookup entirely: {candidate_plants - set(per_plant['Plant Code'])}")

conversion_lookup = (
    per_plant.set_index('Plant Code')
             .reindex(sorted(candidate_plants))
             .reset_index()
)

conversion_lookup['order_count'] = (
    conversion_lookup['order_count'].fillna(0).astype(int)
)

conversion_lookup['reliable_sample'] = (
    conversion_lookup['order_count'] >= MIN_ORDERS_FOR_OWN_AVERAGE
)

conversion_lookup['conversion_factor'] = (
    conversion_lookup['own_avg_order_size']
    .where(conversion_lookup['reliable_sample'], network_avg)
)

conversion_lookup['factor_source'] = (
    conversion_lookup['reliable_sample']
    .map({True: 'own history', False: 'network fallback'})
)

conversion_lookup[['Plant Code', 'order_count', 'own_avg_order_size',
                   'conversion_factor', 'factor_source']]
```

```python
ROAD_FREIGHT_RATE_PER_TONNE_KM_INR = 3.78

orders["weight_per_unit"] = orders["Weight"] / orders["Unit quantity"]

volume_weighted_avg = orders["Weight"].sum() / orders["Unit quantity"].sum()
simple_avg = orders["weight_per_unit"].mean()

print(f"Road freight rate: ₹{ROAD_FREIGHT_RATE_PER_TONNE_KM_INR}/tonne-km")
print(f"Simple average: {simple_avg:.6f} kg/unit")
print(f"Volume-weighted average: {volume_weighted_avg:.6f} kg/unit")

AVG_WEIGHT_PER_UNIT_KG = volume_weighted_avg
```

```python
ROAD_CIRCUITY_FACTOR = 1.3

def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    return R * 2 * np.arcsin(np.sqrt(a))

dc_cust_pairs = warehouse_master[
    ['plant_code', 'latitude', 'longitude']
].merge(
    customer_demand[
        ['Customer', 'latitude', 'longitude']
    ],
    how='cross',
    suffixes=('_dc', '_cust')
)

dc_cust_pairs['straight_line_km'] = haversine_km(
    dc_cust_pairs['latitude_dc'],
    dc_cust_pairs['longitude_dc'],
    dc_cust_pairs['latitude_cust'],
    dc_cust_pairs['longitude_cust']
)

dc_cust_pairs['road_distance_km'] = (
    dc_cust_pairs['straight_line_km'] * ROAD_CIRCUITY_FACTOR
).round(1)

print(f"Total pairs: {len(dc_cust_pairs)}")

dc_cust_pairs.head()
```

```python
dc_cust_pairs["freight_cost_per_unit"] = (
    dc_cust_pairs["road_distance_km"]
    * ROAD_FREIGHT_RATE_PER_TONNE_KM_INR
    * (AVG_WEIGHT_PER_UNIT_KG / 1000)
).round(6)

print(
    dc_cust_pairs[
        [
            "plant_code",
            "Customer",
            "road_distance_km",
            "freight_cost_per_unit"
        ]
    ].head()
)

print(f"Minimum freight cost/unit: ₹{dc_cust_pairs['freight_cost_per_unit'].min():.6f}")
print(f"Maximum freight cost/unit: ₹{dc_cust_pairs['freight_cost_per_unit'].max():.6f}")

dc_cust_pairs.to_csv(
    f"{base}/Data/Processed/freight_cost_matrix.csv",
    index=False
)
```

```python
import pulp

DCs = warehouse_master["plant_code"].tolist()
Customers = customer_demand["Customer"].tolist()

fixed_cost = warehouse_master.set_index("plant_code")["annual_fixed_cost"].to_dict()
variable_cost = warehouse_master.set_index("plant_code")["variable_cost_per_unit"].to_dict()
capacity = warehouse_master.set_index("plant_code")["daily_capacity_units"].to_dict()
demand = customer_demand.set_index("Customer")["avg_daily_demand_units"].to_dict()
freight_cost = dc_cust_pairs.set_index(
    ["plant_code", "Customer"]
)["freight_cost_per_unit"].to_dict()

assert len(fixed_cost) == len(DCs)
assert len(variable_cost) == len(DCs)
assert len(capacity) == len(DCs)
assert len(demand) == len(Customers)
assert len(freight_cost) == len(DCs) * len(Customers)

print("All parameter dictionaries verified complete.")

prob = pulp.LpProblem(
    "Distribution_Network_Optimization",
    pulp.LpMinimize
)

y = pulp.LpVariable.dicts("open", DCs, cat="Binary")
x = pulp.LpVariable.dicts(
    "flow",
    [(i, j) for i in DCs for j in Customers],
    lowBound=0,
    cat="Continuous"
)

print(
    f"Decision variables: {len(y)+len(x)} "
    f"({len(y)} binary + {len(x)} continuous)"
)
```

```python
ANNUAL_OPERATING_DAYS = 250

prob += (
    pulp.lpSum(
        fixed_cost[i] * y[i]
        for i in DCs
    )
    +
    pulp.lpSum(
        (
            variable_cost[i] + freight_cost[(i, j)]
        ) * x[(i, j)] * ANNUAL_OPERATING_DAYS
        for i in DCs
        for j in Customers
    )
), "Total_Annual_Cost"

print("Objective function added successfully.")
print(f"Variables: {len(prob.variables())}")
print(f"Constraints: {len(prob.constraints)}")
```

```python
for j in Customers:
    prob += (
        pulp.lpSum(x[(i, j)] for i in DCs) == demand[j],
        f"Demand_{j}"
    )

for i in DCs:
    prob += (
        pulp.lpSum(x[(i, j)] for j in Customers)
        <= capacity[i] * y[i],
        f"Capacity_{i}"
    )

print("Demand and capacity constraints added.")
print(f"Variables: {len(prob.variables())}")
print(f"Constraints: {len(prob.constraints)}")
```

```python
solver = pulp.PULP_CBC_CMD(msg=True)

result = prob.solve(solver)

print("Status:", pulp.LpStatus[result])
print(f"Objective value: ₹{pulp.value(prob.objective):,.2f}")

open_dcs = [i for i in DCs if pulp.value(y[i]) > 0.5]

print(f"Open DCs ({len(open_dcs)}):")
print(open_dcs)
```

```python
import pandas as pd
import pulp
import os

warehouse_master = pd.read_csv(f"{base}/Data/Processed/warehouse_master.csv")
customer_demand = pd.read_csv(f"{base}/Data/Processed/customer_demand.csv")
freight = pd.read_csv(f"{base}/Data/Processed/freight_cost_matrix.csv")

ANNUAL_OPERATING_DAYS = 250
MIN_SHIPMENT_CHARGE_INR = 1500
AVG_UNITS_PER_SHIPMENT = 3203
min_charge_per_unit = MIN_SHIPMENT_CHARGE_INR / AVG_UNITS_PER_SHIPMENT
SLA_DISTANCE_LIMIT_KM = 600

DCs = warehouse_master["plant_code"].tolist()
Customers = customer_demand["Customer"].tolist()

fixed_cost = warehouse_master.set_index("plant_code")["annual_fixed_cost"].to_dict()
variable_cost = warehouse_master.set_index("plant_code")["variable_cost_per_unit"].to_dict()
capacity = warehouse_master.set_index("plant_code")["daily_capacity_units"].to_dict()
demand = customer_demand.set_index("Customer")["avg_daily_demand_units"].to_dict()
freight_cost = freight.set_index(["plant_code", "Customer"])["freight_cost_per_unit"].to_dict()
distance = freight.set_index(["plant_code", "Customer"])["road_distance_km"].to_dict()

total_demand = sum(demand.values())

def solve_network(sla_multiplier):

    if sla_multiplier is None:
        sla_penalty = {(i, j): 0 for i in DCs for j in Customers}
    else:
        sla_penalty = {
            (i, j): sla_multiplier
            * min_charge_per_unit
            * max(0, distance[(i, j)] - SLA_DISTANCE_LIMIT_KM)
            / 1000
            for i in DCs
            for j in Customers
        }

    prob = pulp.LpProblem("Network", pulp.LpMinimize)

    y = pulp.LpVariable.dicts("open", DCs, cat="Binary")
    x = pulp.LpVariable.dicts(
        "flow",
        [(i, j) for i in DCs for j in Customers],
        lowBound=0,
        cat="Continuous",
    )

    prob += (
        pulp.lpSum(fixed_cost[i] * y[i] for i in DCs)
        + pulp.lpSum(
            (variable_cost[i] + freight_cost[(i, j)] + sla_penalty[(i, j)])
            * x[(i, j)]
            * ANNUAL_OPERATING_DAYS
            for i in DCs
            for j in Customers
        )
    )

    for j in Customers:
        prob += pulp.lpSum(x[(i, j)] for i in DCs) == demand[j]

    for i in DCs:
        prob += (
            pulp.lpSum(x[(i, j)] for j in Customers)
            <= capacity[i] * y[i]
        )

    prob.solve(pulp.PULP_CBC_CMD(msg=0))

    selected = sorted([i for i in DCs if y[i].value() == 1])

    true_cost = (
        sum(
            (variable_cost[i] + freight_cost[(i, j)])
            * x[(i, j)].value()
            * ANNUAL_OPERATING_DAYS
            for i in DCs
            for j in Customers
        )
        + sum(fixed_cost[i] * y[i].value() for i in DCs)
    )

    within_2day = sum(
        x[(i, j)].value()
        for i in DCs
        for j in Customers
        if distance[(i, j)] <= SLA_DISTANCE_LIMIT_KM
    )

    cust_within = sum(
        1
        for j in Customers
        if sum(
            x[(i, j)].value()
            for i in DCs
            if distance[(i, j)] > SLA_DISTANCE_LIMIT_KM
        )
        < 1
    )

    return {
        "sla_multiplier": "baseline" if sla_multiplier is None else sla_multiplier,
        "status": pulp.LpStatus[prob.status],
        "n_dcs_open": len(selected),
        "selected_dcs": ", ".join(selected),
        "true_annual_cost": round(true_cost, 2),
        "cost_vs_baseline_pct": None,
        "sla_volume_pct": round(within_2day / total_demand * 100, 1),
        "sla_customers_pct": round(cust_within / len(Customers) * 100, 1),
    }

sweep_points = [None, 0.5, 1, 3, 5, 8, 12, 15, 18, 20, 25]

results = [solve_network(m) for m in sweep_points]

results_df = pd.DataFrame(results)

baseline_cost = results_df.loc[
    results_df["sla_multiplier"] == "baseline",
    "true_annual_cost",
].iloc[0]

results_df["cost_vs_baseline_pct"] = (
    (results_df["true_annual_cost"] - baseline_cost)
    / baseline_cost
    * 100
).round(2)

print(results_df)

os.makedirs(f"{base}/Outputs/disruption", exist_ok=True)

results_df.to_csv(
    f"{base}/Outputs/disruption/sla_sensitivity_results.csv",
    index=False,
)

print("\nResults saved successfully.")
```

```python
import pandas as pd
import pulp
base = "/content/drive/MyDrive/Distribution_Network_Optimization"
warehouse_master = pd.read_csv(f"{base}/Data/Processed/warehouse_master.csv")
customer_demand = pd.read_csv(f"{base}/Data/Processed/customer_demand.csv")
freight = pd.read_csv(f"{base}/Data/Processed/freight_cost_matrix.csv")
ANNUAL_OPERATING_DAYS = 250
MIN_CHARGE_PER_UNIT = 1500 / 3203
DCs = warehouse_master["plant_code"].tolist()
Customers = customer_demand["Customer"].tolist()
fixed_cost = warehouse_master.set_index("plant_code")["annual_fixed_cost"].to_dict()
variable_cost = warehouse_master.set_index("plant_code")["variable_cost_per_unit"].to_dict()
capacity = warehouse_master.set_index("plant_code")["daily_capacity_units"].to_dict()
demand = customer_demand.set_index("Customer")["avg_daily_demand_units"].to_dict()
weight_based = freight.set_index(
    ["plant_code", "Customer"]
)["weight_based_cost_per_unit"].to_dict()
def solve_fuel_scenario(spike):

    freight_cost = {
        (i, j): MIN_CHARGE_PER_UNIT +
        weight_based[(i, j)] * (1 + spike / 100)
        for i in DCs
        for j in Customers
    }

    prob = pulp.LpProblem("FuelScenario", pulp.LpMinimize)

    y = pulp.LpVariable.dicts("open", DCs, cat="Binary")
    x = pulp.LpVariable.dicts(
        "flow",
        [(i, j) for i in DCs for j in Customers],
        lowBound=0
    )

    prob += (
        pulp.lpSum(fixed_cost[i] * y[i] for i in DCs)
        +
        pulp.lpSum(
            (variable_cost[i] + freight_cost[(i, j)])
            * x[(i, j)]
            * ANNUAL_OPERATING_DAYS
            for i in DCs
            for j in Customers
        )
    )

    for j in Customers:
        prob += pulp.lpSum(x[(i, j)] for i in DCs) == demand[j]

    for i in DCs:
        prob += pulp.lpSum(x[(i, j)] for j in Customers) <= capacity[i] * y[i]

    prob.solve(pulp.PULP_CBC_CMD(msg=0))

    selected = sorted(
        [i for i in DCs if y[i].value() == 1]
    )

    cost = pulp.value(prob.objective)

    return selected, cost

baseline_dcs, baseline_cost = solve_fuel_scenario(0)

fuel_results = []

for spike in [0, 4, 10, 25, 60]:

    selected, cost = solve_fuel_scenario(spike)

    fuel_results.append({
        "fuel_spike_pct": spike,
        "selected_dcs": ", ".join(selected),
        "annual_cost": round(cost, 2),
        "cost_change_pct": round(
            (cost - baseline_cost) / baseline_cost * 100,
            3
        ),
        "network_changed": selected != baseline_dcs
    })

fuel_df = pd.DataFrame(fuel_results)

print(fuel_df)

fuel_df.to_csv(
    f"{base}/Data/Processed/fuel_sensitivity_results.csv",
    index=False
)

print("Results saved successfully.")
```

```python
def solve_with_failure(failed_dc):

    available_dcs = [dc for dc in DCs if dc != failed_dc]

    prob = pulp.LpProblem("DC_Failure", pulp.LpMinimize)

    y = pulp.LpVariable.dicts("open", available_dcs, cat="Binary")

    x = pulp.LpVariable.dicts(
        "flow",
        [(i, j) for i in available_dcs for j in Customers],
        lowBound=0,
        cat="Continuous"
    )

    prob += (
        pulp.lpSum(fixed_cost[i] * y[i] for i in available_dcs)
        +
        pulp.lpSum(
            (variable_cost[i] + freight_cost[(i, j)])
            * x[(i, j)]
            * ANNUAL_OPERATING_DAYS
            for i in available_dcs
            for j in Customers
        )
    )

    for j in Customers:
        prob += pulp.lpSum(x[(i, j)] for i in available_dcs) == demand[j]

    for i in available_dcs:
        prob += pulp.lpSum(x[(i, j)] for j in Customers) <= capacity[i] * y[i]

    prob.solve(pulp.PULP_CBC_CMD(msg=0))

    selected = sorted(
        [i for i in available_dcs if y[i].value() == 1]
    )

    cost = pulp.value(prob.objective)

    return selected, cost
```

```python
failure_results = []

for failed in ["PLANT03", "PLANT04", "PLANT07"]:

    selected, cost = solve_with_failure(failed)

    failure_results.append({
        "failed_dc": failed,
        "replacement_network": ", ".join(selected),
        "annual_cost": round(cost, 2),
        "cost_increase_pct": round(
            (cost - baseline_cost) / baseline_cost * 100,
            2
        )
    })

failure_df = pd.DataFrame(failure_results)

failure_df
```

```python
import pandas as pd
import pulp

base="/content/drive/MyDrive/Distribution_Network_Optimization"

warehouse_master=pd.read_csv(f"{base}/Data/Processed/warehouse_master.csv")
customer_demand=pd.read_csv(f"{base}/Data/Processed/customer_demand.csv")
freight=pd.read_csv(f"{base}/Data/Processed/freight_cost_matrix.csv")
```

```python
ANNUAL_OPERATING_DAYS=250

DCs=warehouse_master["plant_code"].tolist()
Customers=customer_demand["Customer"].tolist()

fixed_cost=warehouse_master.set_index("plant_code")["annual_fixed_cost"].to_dict()
variable_cost=warehouse_master.set_index("plant_code")["variable_cost_per_unit"].to_dict()
capacity=warehouse_master.set_index("plant_code")["daily_capacity_units"].to_dict()
demand=customer_demand.set_index("Customer")["avg_daily_demand_units"].to_dict()
freight_cost=freight.set_index(["plant_code","Customer"])["freight_cost_per_unit"].to_dict()
```

```python
prob=pulp.LpProblem("Distribution_Network_Optimization",pulp.LpMinimize)

y=pulp.LpVariable.dicts("open",DCs,cat="Binary")
x=pulp.LpVariable.dicts("flow",[(i,j) for i in DCs for j in Customers],lowBound=0)

prob+=(
    pulp.lpSum(fixed_cost[i]*y[i] for i in DCs)+
    pulp.lpSum(
        (variable_cost[i]+freight_cost[(i,j)])
        *x[(i,j)]
        *ANNUAL_OPERATING_DAYS
        for i in DCs
        for j in Customers
    )
)

for j in Customers:
    prob+=pulp.lpSum(x[(i,j)] for i in DCs)==demand[j]

for i in DCs:
    prob+=pulp.lpSum(x[(i,j)] for j in Customers)<=capacity[i]*y[i]

prob.solve(pulp.PULP_CBC_CMD(msg=0))

print(pulp.LpStatus[prob.status])
```

```python
rows=[]

for i in DCs:
    for j in Customers:
        if x[(i,j)].value()>0.01:
            rows.append({
                "plant_code":i,
                "Customer":j,
                "daily_units":round(x[(i,j)].value(),1),
                "freight_cost_per_unit":freight_cost[(i,j)],
                "variable_cost_per_unit":variable_cost[i],
                "annual_units":round(x[(i,j)].value()*ANNUAL_OPERATING_DAYS,1)
            })

shipment_plan=pd.DataFrame(rows)

shipment_plan.head()
```
